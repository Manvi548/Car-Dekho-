```python
Q1 = dp["selling_price"].quantile(0.25)
Q3 = dp["selling_price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

dp = dp[(dp["selling_price"] >= lower) & (dp["selling_price"] <= upper)]
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

dp["seller_type"] = le.fit_transform(dp["seller_type"])
dp["fuel_type"] = le.fit_transform(dp["fuel_type"])
dp["transmission_type"] = le.fit_transform(dp["transmission_type"])
```

```python
plt.scatter(dp.mileage,dp.selling_price)
plt.xlabel("mileage")
plt.ylabel("selling_price")
plt.show()
```

```python
x=dp.drop(["selling_price","car_name","brand","model"],axis=1)
y=dp.selling_price
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
model.score(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("Model Evaluaation:")
print(f"R² Score: {r2:.3f}")
print(f"Mean Absolute Error: {mae:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
```
